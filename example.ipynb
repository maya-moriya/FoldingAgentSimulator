# FoldingAgent Simulator — Usage Example

Build a piece of virtual paper, add vertices where new creases are needed, and fold along them step by step. Each `plot()` call renders the current state of the model.

```python
from origami import Origami

origami = Origami(front_color='brown', back_color='lightyellow')
origami.plot()

origami.rotate(45)
origami.plot()

origami.fold((2, 4), 1)
origami.plot()

origami.add_vertex((1, 4), 0.6)
origami.add_vertex((1, 2), 0.6)
origami.fold((6, 5), -1)
origami.plot()

origami.flip('x')
origami.plot()

origami.add_vertex((1, 5), 0.333333333)
origami.add_vertex((1, 6), 0.333333333)
origami.fold((7, 8), 1)
origami.plot()

origami.flip('x')
origami.plot()

origami.add_vertex((7, 8), 0.5)
origami.fold((9, 3), 1)
origami.plot()

origami.fold((6, 9), 1)
origami.plot()

origami.add_vertex((4, 5), 0.8)
origami.fold((9, 14), 1)
origami.plot()

origami.flip('x')
origami.plot()

origami.add_vertex((14, 11), 0.3)
origami.add_vertex((17, 4), 0.3)
origami.fold((18, 19), 1)
origami.plot()

origami.flip('x')
origami.plot()
```

## Exporting

Save the final folded model and its crease pattern to `.png` files.

```python
origami.plot(debug=False, save_path='example_render.png')
origami.plot_cp(save_path='example_cp.png')
```

The model can also be exported to the [FOLD](https://github.com/edemaine/fold) interchange format and written to disk.

```python
origami.export_cp(save_path='example.fold')
```
